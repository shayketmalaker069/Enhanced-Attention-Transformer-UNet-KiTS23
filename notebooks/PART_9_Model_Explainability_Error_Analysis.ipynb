```python

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_PATH = r"C:\Users\USER\Kidney_Segmentation_Project"

RESULT_PATH = os.path.join(PROJECT_PATH,"results")
FIGURE_PATH = os.path.join(PROJECT_PATH,"figures")

os.makedirs(FIGURE_PATH, exist_ok=True)

print("Environment ready")
```

```python

def calculate_error_maps(prediction, ground_truth):

    false_positive = np.logical_and(
        prediction > 0,
        ground_truth == 0
    )

    false_negative = np.logical_and(
        prediction == 0,
        ground_truth > 0
    )

    true_positive = np.logical_and(
        prediction > 0,
        ground_truth > 0
    )

    return false_positive, false_negative, true_positive
```

```python

def error_visualization(ct, fp, fn, tp, filename):

    plt.figure(figsize=(15,5))

    plt.subplot(1,3,1)
    plt.imshow(ct,cmap="gray")
    plt.title("False Positive Map")
    plt.imshow(fp,alpha=0.5)
    plt.axis("off")

    plt.subplot(1,3,2)
    plt.imshow(ct,cmap="gray")
    plt.title("False Negative Map")
    plt.imshow(fn,alpha=0.5)
    plt.axis("off")

    plt.subplot(1,3,3)
    plt.imshow(ct,cmap="gray")
    plt.title("Correct Segmentation")
    plt.imshow(tp,alpha=0.5)
    plt.axis("off")

    plt.savefig(
        os.path.join(FIGURE_PATH,filename),
        dpi=600,
        bbox_inches="tight"
    )

    plt.show()

print("Error visualization ready")
```

```python

def calculate_error_metrics(pred,true):

    fp=np.sum((pred>0)&(true==0))
    fn=np.sum((pred==0)&(true>0))
    tp=np.sum((pred>0)&(true>0))

    sensitivity=tp/(tp+fn+1e-8)
    specificity=1-(fp/(fp+tp+1e-8))

    return {
        "False Positive":fp,
        "False Negative":fn,
        "Sensitivity":sensitivity,
        "Specificity":specificity
    }

print("Metrics function ready")
```
