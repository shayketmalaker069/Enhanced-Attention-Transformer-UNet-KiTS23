```python

import os
import numpy as np
import matplotlib.pyplot as plt

PROJECT_PATH=r"C:\Users\USER\Kidney_Segmentation_Project"

FIGURE_PATH=os.path.join(PROJECT_PATH,"figures")

os.makedirs(FIGURE_PATH,exist_ok=True)
```

```python

def create_segmentation_overlay(ct,mask,title,path):

    overlay=np.zeros(
        (mask.shape[0],mask.shape[1],3)
    )

    overlay[mask==1]=[0,1,0]
    overlay[mask==2]=[1,0,0]

    plt.figure(figsize=(6,6),dpi=300)

    plt.imshow(ct,cmap="gray")
    plt.imshow(overlay,alpha=0.45)

    plt.title(title)
    plt.axis("off")

    plt.savefig(
        path,
        dpi=600,
        bbox_inches="tight"
    )

    plt.show()
```
