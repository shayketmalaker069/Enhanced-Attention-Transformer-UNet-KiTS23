```python

import os
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_PATH=r"C:\Users\USER\Kidney_Segmentation_Project"

RESULT_PATH=os.path.join(PROJECT_PATH,"results")
FIGURE_PATH=os.path.join(PROJECT_PATH,"figures")

os.makedirs(FIGURE_PATH,exist_ok=True)
```

```python

final_table=pd.DataFrame({
"Model":[
"3D U-Net",
"nnU-Net",
"Enhanced Attention Transformer U-Net"
],
"Dice":[0,0,0],
"IoU":[0,0,0],
"Precision":[0,0,0],
"Recall":[0,0,0],
"HD95":[0,0,0]
})

final_table
```

```python

final_table.to_csv(
    os.path.join(
        RESULT_PATH,
        "Final_Model_Comparison.csv"
    ),
    index=False
)
```

```python

plt.figure(figsize=(9,5))

plt.bar(
    final_table["Model"],
    final_table["Dice"]
)

plt.ylabel("Dice Score")
plt.title("Final Model Comparison")

plt.xticks(rotation=20)
plt.ylim(0,1)
plt.grid(axis="y")

plt.savefig(
    os.path.join(
        FIGURE_PATH,
        "Final_Model_Dice_Comparison.png"
    ),
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```
