# PART 6 — Advanced Evaluation + Statistical Analysis

This notebook performs:
- Load all model results
- Statistical summary
- Mean and standard deviation analysis
- Model improvement percentage
- Tumour-focused analysis
- Comparison graphs


```python

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

PROJECT_PATH = r"C:\Users\USER\Kidney_Segmentation_Project"
RESULT_PATH = os.path.join(PROJECT_PATH,"results")
FIGURE_PATH = os.path.join(PROJECT_PATH,"figures")

os.makedirs(FIGURE_PATH,exist_ok=True)
```

```python

# Load results files

files = {
    "3D U-Net":"3D_UNet_results.csv",
    "nnU-Net":"nnUNet_results.csv",
    "Enhanced Model":"Enhanced_Model_results.csv"
}

all_results = {}

for name,file in files.items():
    path=os.path.join(RESULT_PATH,file)
    if os.path.exists(path):
        all_results[name]=pd.read_csv(path)

print(all_results.keys())
```

```python

# Statistical summary

summary=[]

for model,data in all_results.items():
    summary.append({
        "Model":model,
        "Mean Dice":data["Dice"].mean(),
        "STD Dice":data["Dice"].std(),
        "Mean IoU":data["IoU"].mean(),
        "STD IoU":data["IoU"].std(),
        "Mean HD95":data["HD95"].mean()
    })

summary_df=pd.DataFrame(summary)
summary_df
```

```python

# Save statistical table

summary_df.to_csv(
    os.path.join(RESULT_PATH,"Statistical_Analysis.csv"),
    index=False
)
```

```python

# Tumour Dice comparison

values=[]

for model,data in all_results.items():
    values.append(data["Tumour Dice"].mean())

plt.figure(figsize=(8,5))
plt.bar(all_results.keys(),values)
plt.ylabel("Tumour Dice")
plt.title("Tumour Segmentation Performance Comparison")
plt.ylim(0,1)
plt.xticks(rotation=20)
plt.grid(axis="y")

plt.savefig(
    os.path.join(FIGURE_PATH,"Tumour_Dice_Analysis.png"),
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```
