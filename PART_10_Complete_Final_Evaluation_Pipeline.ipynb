## 3D Kidney and Kidney Tumor Segmentation Using Deep Learning


```python

# ==========================================
# 1. IMPORT LIBRARIES
# ==========================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.spatial.distance import directed_hausdorff

PROJECT_PATH = r"C:\Users\USER\Kidney_Segmentation_Project"

RESULT_PATH = os.path.join(PROJECT_PATH,"results")
FIGURE_PATH = os.path.join(PROJECT_PATH,"figures")

os.makedirs(RESULT_PATH,exist_ok=True)
os.makedirs(FIGURE_PATH,exist_ok=True)

print("Environment Ready")
```

```python

# ==========================================
# 2. METRIC FUNCTIONS
# ==========================================

def dice_score(pred,true):

    intersection = np.sum(pred * true)

    return (
        2*intersection /
        (
            np.sum(pred)+
            np.sum(true)+
            1e-8
        )
    )


def iou_score(pred,true):

    intersection=np.logical_and(
        pred,true
    ).sum()

    union=np.logical_or(
        pred,true
    ).sum()

    return intersection/(union+1e-8)


def precision_score(pred,true):

    tp=np.logical_and(
        pred,true
    ).sum()

    fp=np.logical_and(
        pred,
        np.logical_not(true)
    ).sum()

    return tp/(tp+fp+1e-8)


def recall_score(pred,true):

    tp=np.logical_and(
        pred,true
    ).sum()

    fn=np.logical_and(
        np.logical_not(pred),
        true
    ).sum()

    return tp/(tp+fn+1e-8)


print("Dice, IoU, Precision, Recall functions created")
```

```python

# ==========================================
# 3. HD95 BOUNDARY METRIC
# ==========================================

def hd95(pred,true):

    pred_points=np.argwhere(pred)

    true_points=np.argwhere(true)

    if len(pred_points)==0 or len(true_points)==0:
        return np.nan


    forward=directed_hausdorff(
        pred_points,
        true_points
    )[0]


    backward=directed_hausdorff(
        true_points,
        pred_points
    )[0]


    return np.percentile(
        [forward,backward],
        95
    )


print("HD95 function ready")
```

```python

# ==========================================
# 4. COMPLETE MODEL EVALUATION FUNCTION
# ==========================================

def evaluate_segmentation(prediction, ground_truth):

    result={}

    result["Dice"] = dice_score(
        prediction>0,
        ground_truth>0
    )

    result["IoU"] = iou_score(
        prediction>0,
        ground_truth>0
    )

    result["Precision"] = precision_score(
        prediction>0,
        ground_truth>0
    )

    result["Recall"] = recall_score(
        prediction>0,
        ground_truth>0
    )

    result["HD95"] = hd95(
        prediction>0,
        ground_truth>0
    )


    # Kidney

    result["Kidney Dice"] = dice_score(
        prediction==1,
        ground_truth==1
    )


    # Tumour

    result["Tumour Dice"] = dice_score(
        prediction==2,
        ground_truth==2
    )


    return result


print("Complete evaluation function ready")
```
